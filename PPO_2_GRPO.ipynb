## PPO
在`PPO.py`里, 每个action后, environment都会给一个reward. 但是在LLM Post training里, 只有整个输出完成后, 才会给一个最终的reward (是episode reward, 而非step reward).

## GRPO（Group Relative Policy Optimization)
> While in the LLM context, **usually only the last token is assigned a
reward score by the reward model, which may complicate the training of a value function that is accurate at each token**. To address this, as shown in Figure 4, we propose Group Relative Policy Optimization (GRPO), which obviates the need for additional value function approximation as
in PPO, and instead uses the average reward of multiple sampled outputs, produced in response
to the same question, as the baseline.

![](./figures/PPO_GRPO.png)

可以看到, PPO是通过GAE来估计advantage, 但是GRPO将value model去掉了, 通过group computation直接从reward得到advantage。具体的advantage计算方式为: 对每个quesiton prompt $q$, 我们可以采样得到G个output sequence ${o_1, o_2, ..., o_G}$。 通过reward model, 可以得到G个reward score ${r_1, r_2, ..., r_G}$。 最后通过减去均值，然后除以标准差来进行标准化。
$$\hat A_{i,t} = r_i = \frac{r_i - \text{mean}(\mathbf{r})}{\text{std}(\mathbf{r})}$$
