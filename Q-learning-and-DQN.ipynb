## 1. Model-Free Value function prediction

$V(s)$: state value function, $V_{t}(s) = \mathbf{E}[G_t|S_t=s] =\mathbf{E}[R_{t} + \gamma V_{t}(S)]$ 表示$t$时刻处于状态$s$的期望回报(return: sum of discounted reward)



### 1.1 蒙特卡洛法估计V(s)

- 选取状态$s$被访问到的一些episode
- 使用$N(s)$统计状态$s$被访问到的次数
- 增量计算每次访问$s$带来的回报: $S(s) = S(s) + G_t$
- 根据蒙特卡洛思想, 当采样的episode足够多时, 可以用$V(s) =\frac{S(s)}{N(s)}$来估计真实的sate value function $V(s)$

![](./figures/MC_Value_Estimation.png)



### 1.2 TD learning估计V(s)

TD: Temporal difference (时序差分)结合了蒙特卡洛和动态规划方法:
- 类似蒙特卡洛法, 也是从实际episode中学习
- 类似动态规划法, 通过bootstrapping(自举)从incomplte episode学习

在蒙特卡罗法里:
$$G_t = R_{t} + \gamma R_{t+1} + \gamma ^2 R_{t+2} + ...$$


在$n$-step TD learning里:

$$G_t = R_{t} + \gamma R_{t+1} + ... + \gamma^{n-1} R_{t+n-1} + \color{red}{\gamma^{n}V(S_{t+n})}$$

如果是1-step TD learning算法, 就是用当前获得的奖励$R_{t+1}$加上下一个状态的价值估计来作为当前状态回获得的回报$G_t$

所以基于TD learning的state value function的迭代公式为:

$$V(S_t) \leftarrow V(S_t) + \alpha \underbrace{\color{red}{[R_{t} + \gamma V(S_{t+1}) - V(S_t)]}}_{\text{TD Error}}$$

## SARSA and Q-Learning
既然可以使用**时序差分**方法来**估计**价值函数, 那么也可以用这一方法来估计Q函数(state-action value function).

如果再配合一定的策略来选取某个特定action, 就可以形成一个完整的强化学习算法. 这一节介绍SARSA和Q-Learning这两个**基于时序差分**的强化学习算法


### SARSA

SARSA是一种on-policy的TD control算法:
![](./figures/SARSA.png)

- 行为策略: 用于采样数据的策略
- 目标策略： 用采样数据来更新的策略

on-policy: 行为策略和目标策略是同一个策略

### Q-Learning
Q-Learning是一种off-policy算法, 不同于SARSA必须使用来自当前策略采样得到的五元组$(s,a,r,s',a')$来更新, Q-learning的更新公式使用的是四元组$(s,a,r,s')$来更新$Q(s,a)$, 而且该四元组并不需要一定是当前策略采样得到的数据，也可以来自行为策略
![](./figures/Q-Learning.png)
